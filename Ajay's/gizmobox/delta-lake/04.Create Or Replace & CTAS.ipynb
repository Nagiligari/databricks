```python
%sql
DROP TABLE IF EXISTS gizmobox.demo.compaines;
```

```python
%sql

CREATE OR REPLACE TABLE gizmobox.demo.compaines(
    company_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
    company_name STRING NOT NULL,
    founded_date DATE COMMENT 'The date the company has founded on',
    founded_year INT GENERATED ALWAYS AS (YEAR(founded_date)),
    country STRING
)
COMMENT 'This table contains information about some of the successful tech compaines'
TBLPROPERTIES ('sensitive' = 'true', 'delta.enableDeletionVectors'= 'false');
```

```python
%sql
INSERT INTO gizmobox.demo.compaines (company_name,founded_date,country)
VALUES ("Apple","1976-04-01","USA"),
("Microsoft","1975-04-04","USA"),
("Google","1998-09-04","USA"),
("Amazon","1994-07-05","USA"),
("Tencent","1998-11-11","China")
```

```python
%sql
DESC HISTORY gizmobox.demo.compaines;
```

### CTAS STATEMENT

```python
%sql
CREATE TABLE gizmobox.demo.compaines_china AS
SELECT * FROM gizmobox.demo.compaines WHERE country = 'China';
```

- we cannot specify column properties in ctas statements because column properties are inferred from source table

```python
%sql
describe history gizmobox.demo.compaines_china
```
