```python
%sql
DROP TABLE IF EXISTS gizmobox.demo.gold_companies;

CREATE TABLE gizmobox.demo.gold_companies
(company_name STRING,
founded_date DATE,
country STRING);

INSERT INTO gizmobox.demo.gold_companies
VALUES ('Acme Inc', '2010-01-01', 'USA'),
('Beta Corp', '2015-01-01', 'Canada');

SELECT * FROM gizmobox.demo.gold_companies
```

```python
%sql
INSERT OVERWRITE TABLE gizmobox.demo.gold_companies
SELECT company_name,founded_date,country FROM gizmobox.demo.compaines;
```

```python
%sql
DESCRIBE HISTORY gizmobox.demo.gold_companies
```

```python
%sql
DROP TABLE IF EXISTS gizmobox.demo.gold_companies_partitioned;

CREATE TABLE gizmobox.demo.gold_companies_partitioned
(company_name STRING,
founded_date DATE,
country STRING)
PARTITIONED BY (country);

INSERT INTO gizmobox.demo.gold_companies_partitioned
VALUES ('Acme Inc', '2010-01-01', 'USA'),
('Beta Corp', '2015-01-01', 'Canada');

SELECT * FROM gizmobox.demo.gold_companies_partitioned;
```

```python
%sql
DESCRIBE EXTENDED gizmobox.demo.gold_companies_partitioned
```
