```python
%sql 
DROP TABLE IF EXISTS gizmobox.demo.compaines;
```

```python
%sql

CREATE TABLE gizmobox.demo.compaines(
    company_name STRING NOT NULL,
    founded_date DATE COMMENT 'The date the company has founded on',
    country STRING
)
COMMENT 'This table contains information about some of the successful tech compaines'
TBLPROPERTIES ('sensitive' = 'true', 'delta.enableDeletionVectors'= 'false')
```

```python
%sql

CREATE TABLE gizmobox.demo.compaines(
    company_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
    company_name STRING NOT NULL,
    founded_date DATE COMMENT 'The date the company has founded on',
    country STRING
)
COMMENT 'This table contains information about some of the successful tech compaines'
TBLPROPERTIES ('sensitive' = 'true', 'delta.enableDeletionVectors'= 'false')
```

```python
%sql

CREATE TABLE gizmobox.demo.compaines(
    company_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
    company_name STRING NOT NULL,
    founded_date DATE COMMENT 'The date the company has founded on',
    founded_year INT GENERATED ALWAYS AS (YEAR(founded_date)),
    country STRING
)
COMMENT 'This table contains information about some of the successful tech compaines'
TBLPROPERTIES ('sensitive' = 'true', 'delta.enableDeletionVectors'= 'false')
```
