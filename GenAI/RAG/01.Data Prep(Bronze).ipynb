```python
datasource = '/Volumes/genai/landing/dataset/docs'
```

```python
from pyspark.sql.functions import current_timestamp
docs_raw_df = (
    spark.read.format('binaryfile').load(f'{datasource}')
    .withColumn('ingestion_time',current_timestamp())
)
display(docs_raw_df)
```

```python
docs_raw_df.write.mode("overwrite").saveAsTable("genai.bronze.docs_raw")
```
