```python

data = [
    (1,),
    (0,),
    (1,),
    (None,)
]

columns = ["id"]

df1 = spark.createDataFrame(data,columns)
df1.display()
```

```python

data = [
    (1,),
    (None,),
    (0,),
    (None,)
]

columns = ["id"]

df2 = spark.createDataFrame(data,columns)
df2.display()
```

##### Left join - Returns all rows from the left DataFrame and matching rows from the right. If there is no match, right-side columns become NULL.
1   1
0   0
1   0
null null

```python
left_join_df = df1.alias("tableA").join(df2.alias("tableB"),"id","left")
left_join_df = left_join_df.select("tableA.id","tableB.id")
display(left_join_df)
```

```python
inner_join_df = df1.alias("tableA").join(df2.alias("tableB"),"id","inner")
inner_join_df = inner_join_df.select("tableA.id","tableB.id")
display(inner_join_df)
```

```python
right_join_df = df1.alias("tableA").join(df2.alias("tableB"),"id","right")
right_join_df = right_join_df.select("tableA.id","tableB.id")
display(right_join_df)
```

```python
leftsemi_join_df = df1.alias("tableA").join(df2.alias("tableB"),"id","leftsemi")
leftsemi_join_df = leftsemi_join_df.select("*")
display(leftsemi_join_df)
```

```python
leftanti_join_df = df1.alias("tableA").join(df2.alias("tableB"),"id","leftanti")
leftanti_join_df = leftanti_join_df.select("*")
display(leftanti_join_df)
```
