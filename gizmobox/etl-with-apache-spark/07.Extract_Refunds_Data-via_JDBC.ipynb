if your ADB workspace is created post 18th-Dec-2025, you cannot use hive metastore(i.e., Cell 2) because post dec 2025 databricks is creating UC enabled workspace. To create external loaction using external database. you need to use Lakehouse Federation

```python
%sql
CREATE TABLE IF NOT EXISTS hive_metastore.bronze.refunds
USING JDBC
OPTIONS(
    url "jdbc:sqlserver://gizmoboxsrv1997.database.windows.net:1433;database=gizmobox-db",
    dbtable "refunds",
    user "ggizmoboxadm",
    password "Darling@2603"
);
```

```python
%sql
CREATE CONNECTION asql_gizmobox_db_conn_sql TYPE sqlserver
OPTIONS(
    host "gizmoboxsrv1997.database.windows.net",
    port "1433",
    user "gizmoboxadm",
    password "Darling@2603"
)
```

```python
%sql
CREATE FOREIGN CATALOG IF NOT EXISTS asql_gizmobox_db_catalog_sql USING CONNECTION asql_gizmobox_db_conn_sql
OPTIONS (
    database 'gizmobox-db'
);
```

```python
%sql
SHOW CATALOGS;
USE CATALOG asql_gizmobox_db_catalog_sql;
SHOW SCHEMAS;
USE SCHEMA dbo;
SHOW TABLES;
    
SELECT * FROM refunds;
```

```python
%sql
DROP CATALOG IF EXISTS asql_gizmobox_db_catalog_sql CASCADE;
DROP CONNECTION IF EXISTS asql_gizmobox_db_conn_sql;
```
