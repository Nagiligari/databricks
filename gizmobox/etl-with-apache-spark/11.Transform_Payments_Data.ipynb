### Transform Payments Data
1. Extract Date and Time from payment_timestamp and create new columns payment_date and payment_time
2. Map payment_status to contain descriptive values 
    (1- Success, 2-Pending, 3-Cancelled, 4- Failed)
3. Write transformed data to the silver schema

```python
%sql
select * from gizmobox.bronze.payments
```

```python
%sql
select 
payment_id,
customer_id,
cast(date_format(payment_date, 'yyyy-MM-dd') as date) as payment_date,
date_format(payment_date, 'HH:mm:ss') as payment_time,
payment_status,
payment_method
from gizmobox.bronze.payments
```

```python
%sql
select 
payment_id,
customer_id,
cast(date_format(payment_date, 'yyyy-MM-dd') as date) as payment_date,
date_format(payment_date, 'HH:mm:ss') as payment_time,
CASE payment_status
WHEN 1 THEN 'Success'
WHEN 2 THEN 'Pending'
WHEN 3 THEN 'Cancelled'
WHEN 4 THEN 'Failed'
END as payment_status,
payment_method
from gizmobox.bronze.payments
```

```python
%sql
CREATE TABLE gizmobox.silver.payments
USING DELTA
LOCATION 'abfss://gizmobox@nagiligaristg.dfs.core.windows.net/silver/payments'
AS
select 
payment_id,
customer_id,
cast(date_format(payment_date, 'yyyy-MM-dd') as date) as payment_date,
date_format(payment_date, 'HH:mm:ss') as payment_time,
CASE payment_status
WHEN 1 THEN 'Success'
WHEN 2 THEN 'Pending'
WHEN 3 THEN 'Cancelled'
WHEN 4 THEN 'Failed'
END as payment_status,
payment_method
from gizmobox.bronze.payments
```
