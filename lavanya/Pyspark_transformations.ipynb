```python
df_select =  df.select('price','city','customer_id')
display(df_select)
```

```python
df_alias =  df.select(col("country").alias('customer_country') , col("city").alias("customer_city"))
display(df_alias)
```

```python
df_filters1 =  df.filter(col('order_status')== 'Returned')
display(df_filters1)
```

```python
df_filters2 =  df.filter((col('order_status')== 'Returned') | (col('order_status')== 'Cancelled'))
display(df_filters2)
```

```python
df_filters3 =  df.filter(col('order_status').isin(['Returned','Cancelled','Shipped']))
display(df_filters3)
```

```python
df_rename = df.withColumnRenamed('order_status', 'order_status_info')
display(df_rename)
```

```python
df_column = df.withColumn('flag', lit(0))
display(df_column)
```

```python
df_replaceval = df.withColumn('shipping_address', regexp_replace('shipping_address', ',.*', ""))
display(df_replaceval)
```
