```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df_csv =(spark.read
        .format('CSV')
        .option("header",True)
        .option("inferSchema",True)
        .load('/Volumes/sparkcatalog/landing/data_source/orders.csv')
        )

df_csv.display()
```

```python
df_json = (spark.read
           .format("JSON")
           .option("inferSchema",True)
           .load("/Volumes/sparkcatalog/landing/data_source/orders.json")
           )
display(df_json)
```

```python
df_parquet = (
    spark.read.format('parquet')
    .load("/Volumes/sparkcatalog/landing/data_source/part-00000-tid-orders.c000.snappy.parquet")
)
display(df_parquet)
```

```python
df_permissive = (
    spark
    .read
    .format("JSON")
    .option("inferSchema",True)
    .option("header",True)
    .option("mode","PERMISSIVE")
    .load("/Volumes/sparkcatalog/landing/data_source/orders.json")
)
display(df_permissive)
```
