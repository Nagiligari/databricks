```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
data =[
    (1,"nagiligari","nagiligari@gmail.com"),
    (2,"ajay","ajay@gmail.com"),
    (3,"kumar","kumar@gmail.com")
]
columns = ["id","name","email"]
df_new = spark.createDataFrame(data,columns)
```

```python
df_new = df_new.filter(col("name").like('%i'))
display(df_new)
```

```python
df_new.explain(True)
```
