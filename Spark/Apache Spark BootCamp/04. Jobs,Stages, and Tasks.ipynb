```python
df = (
    spark
    .read
    .format("csv")
    .option("header",True)
    .load("/Volumes/learnspark/raw/spark_volume/raw_orders/")
)
```

```python
# wide Transformation - 1
df = df.repartition(2) # repatition will create dedicated number of partitions that can be processed in parallel. by default it is 200

# Narrow Transformation -1 
df = df.select("order_id","customer_id")

# Narrow Transformation - 2 
df = df.filter(col("order_id") == 1001)

# wide Transformation - 3 
df = df.groupBy("customer_id").count()
```
