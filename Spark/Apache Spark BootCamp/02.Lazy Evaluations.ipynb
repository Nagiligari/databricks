```python
data = [
    (1,"Nagiligari","Nagiligari@gmail.com"),
    (2,"Ajay","Ajay@gmail.com"),
    (3,"kumar","kumar@gmail.com"),
    (4,"suresh","suresh@gmail.com"),
    (5,"Ravi","Ravi@gmail.com"),
    (6,"ani","ani@gmail.com")
]
columns = ["id","name","email"]
df_new = spark.createDataFrame(data,columns)

# Transformation - 1 
df_new = df_new.select("id","email")

#  Transformation - 2
df_new = df_new.filter(col("id") > 3)
```
